```python
import pandas as pd
```

```python
df = pd.read_csv("2. Finance_Analytics_Dataset.csv")
```

```python
#show the number of transaction in each category using count plot
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x='Category', data=df)
plt.title("Number of Transactions in Each Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
#show number of transaction for each payment mode using count plot 
sns.countplot(x="PaymentMode", data=df)
plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
#Show the number of transaction by transaction type(Debit and Credit)using a count plot
sns.countplot(x="TransactionType", data=df)
plt.title("Number of Transactions by Transaction Type")
plt.xlabel("TransactionType")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
# Debit vs Credit Amount Comparsion(bar plot)

sns.barplot(x="TransactionType", y="Amount", data=df)
plt.title("Debit vs Credit Amount Comparison")
plt.xlabel("TransactionType")
plt.ylabel("Amount")
plt.show()
```

```python
#Spending pattern by customer segment(Bar plot)
sns.barplot(x="CustomerSegment", y="Amount", data=df)
plt.title("Spending Pattern by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Spending")
plt.show()
```

```python
#Payment mode usage Distribution(piechart)
counts = df["PaymentMode"].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Payment Mode Usage")
plt.show()
```

```python
#boxplot of transaction Amount by category
sns.boxplot(x='Category', y='Amount', data=df)
plt.title("Transaction Amount by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()
```
