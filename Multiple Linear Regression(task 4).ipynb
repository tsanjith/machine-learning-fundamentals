```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('house_price.csv')
df.head()
```

```python
X = df[['House Size', 'Bedrooms', 'Age of House']]
Y = df['House Price']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
Y_pred
```

```python
r2_score_value = r2_score(Y_test, Y_pred)
r2_score_value
print(f"R-squared Score: {r2_score_value}")
```

```python
plt.scatter(Y_test, Y_pred, color='blue', label='Actual vs Predicted')
plt.xlabel('Actual House Price ($)')
plt.ylabel('Predicted House Price ($)')
plt.title('Actual vs Predicted House Price')
plt.legend()
plt.show()
```

```python


user_House_Size = int(input("Enter House Size: "))
user_Bedrooms = int(input("Enter Number of Bedrooms: "))
user_Age = int(input("Enter Age of House: "))
user_df = pd.DataFrame([[user_House_Size, user_Bedrooms, user_Age]], columns=['House Size', 'Bedrooms', 'Age of House'])
predicted_price = model.predict(user_df)
print(f"The predicted House Price is: ${predicted_price[0]:.2f}")
```
