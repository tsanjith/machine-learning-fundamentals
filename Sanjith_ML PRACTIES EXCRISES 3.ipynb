```python
import pandas as pd
```

```python
df = pd.read_csv("customer_orders.csv")
df
```

```python
df['CustomerName'] = df['CustomerName'].fillna('Unknown')
```

```python
df['Product'] = df['Product'].fillna('Unknown Product')
```

```python
df = df.dropna(subset=['Quantity'])
```

```python
mean_price = df['PricePerUnit'].mean()
df['PricePerUnit'] = df['PricePerUnit'].fillna(mean_price)

print(df)
```

```python
invalid_values = df[df['Quantity'] < 0]
```

```python
print("Invalid values:")
print(invalid_values)
```

```python
df.loc[df['Quantity'] < 0, 'Quantity'] = df.loc[df['Quantity'] < 0, 'Quantity'].abs()
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()
```

```python
df['Quantity'] = df['Quantity'].astype(int)

print(df.dtypes)
```

```python
df['TotalPrice'] = df['Quantity'] * df['PricePerUnit']

print(df)
```
