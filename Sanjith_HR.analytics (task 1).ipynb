```python
import pandas as pd
```

```python
df = pd.read_csv("hr_analytics.csv")
```

```python
print(df.shape)
```

```python
rows, columns = df.shape
print("Number of rows:", rows)
print("Number of columns:", columns)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum().sum())
```

```python
print(df.duplicated().any())
```

```python
print("Number of employees:", df.shape[0])
```

```python
num_departments = df['Department'].nunique()
print("Number of departments:", num_departments)
```

```python
#  Display all departments
departments = df['Department'].unique()
print("Departments:", departments)
```

```python
#  Average employee age
avg_age = df['Age'].mean()
print("Average Age:", avg_age)
```

```python
# Minimum and Maximum employee age
min_age = df['Age'].min()
max_age = df['Age'].max()
print("Minimum Age:", min_age)
print("Maximum Age:", max_age)
```

```python
# Distribution of employees by Department
print(df['Department'].value_counts())
```

```python
# Count employees in each department
dept_counts = df['Department'].value_counts()
print(dept_counts)
```

```python
# Count employees with more than 10 years experience
count = df[df['YearsAtCompany'] > 10].shape[0]
print("Employees with >10 years experience:", count)
```

```python
# Count employees with > 5 years
count_5yrs = (df['YearsAtCompany'] > 5).sum()
print("Employees with >5 years:", count_5yrs)
```

```python
# Average Monthly Income per Department
avg_income_dept = df.groupby('Department')['MonthlyIncome'].mean()
print(avg_income_dept)
```

```python
# Count employees who left (Attrition = Yes) in each department
left_counts = df[df['Attrition'] == 'Yes'].groupby('Department').size()
print(left_counts)
```
